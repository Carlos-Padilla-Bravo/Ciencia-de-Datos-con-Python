```python

# Graficando porcentaje de clientes por ciudad

# Configurar el estilo de ggplot para un look más profesional
plt.style.use('ggplot')

# Calcula el porcentaje para cada ciudad
porcentaje_ciudad = df_clientes['ciudad'].value_counts(normalize=True) * 100

# --- Optimizaciones para un look profesional ---

# 1. Elige una paleta de colores profesional
colors = plt.cm.viridis(np.linspace(0, 1, len(porcentaje_ciudad)))

# 2. "Explota" la porción más grande para darle énfasis
explode = [0] * len(porcentaje_ciudad)
if not porcentaje_ciudad.empty:
    explode[0] = 0.1  # Destaca la primera porción (la más grande)

# 3. Crea la figura y los ejes
fig, ax = plt.subplots(figsize=(8, 5))

# 4. Dibuja el gráfico de pastel con las mejoras
wedges, texts, autotexts = ax.pie(porcentaje_ciudad,
                                      autopct='%1.1f%%',
                                      startangle=140,
                                      colors=colors,
                                      explode=explode,
                                      shadow=False,
                                      pctdistance=0.85) # Distancia del texto de porcentaje al centro

# Mejora las propiedades del texto dentro del pastel
plt.setp(autotexts, size=10, weight="bold", color="black")

# 5. Asegura que el gráfico sea un círculo
ax.axis('equal')

# 6. Añade un título claro
ax.set_title("Porcentaje de Clientes por Ciudad", fontsize=16, pad=20, fontweight='bold')

# 7. Gestiona la leyenda
ax.legend(wedges, porcentaje_ciudad.index,
              title="Ciudades",
              loc="center left",
              bbox_to_anchor=(1, 0, 0.5, 1))

# Ajusta el layout para hacer espacio a la leyenda
plt.tight_layout(rect=[0, 0, 0.85, 1])
plt.show()
```

```python
# Recategorizando los productos basándonos en un set predefinido de productos de limpieza

# 1. Set de productos de limpieza identificado en el data frame (17 productos)
LISTA_LIMPIEZA = {
    "Esponjas x3", "Limpiavidrios 500ml", "Mascarilla Capilar",
    "Cepillo de Dientes", "Desodorante Aerosol", "Servilletas x100",
    "Shampoo 400ml", "Lavandina 1L", "Trapo de Piso",
    "Desengrasante 500ml", "Suavizante 1L", "Hilo Dental",
    "Crema Dental 90g", "Toallas Húmedas x50", "Papel Higiénico x4",
    "Jabón de Tocador", "Detergente Líquido 750ml"
}

# 2. Función de categorización
def categorize_by_list(nombre_producto):
    if nombre_producto in LISTA_LIMPIEZA:
        return 'Limpieza'
    else:
        return 'Alimentos'

# 3. Creando un nuevo data frame (copia) y aplicando la función para crear la nueva columna en tu DataFrame

df_productos_nuevo = df_productos.copy()
df_productos_nuevo['categoria_corregida'] = df_productos_nuevo['nombre_producto'].apply(categorize_by_list)

print("¡Columna 'categoria_corregida' añadida!")

# 4. Agrupando los productos por la nueva categoría para verificar resultados
categoria_counts = df_productos_nuevo['categoria_corregida'].value_counts()
print(categoria_counts)
```

```python
# Creando un nuevo data frame de productos sin la categoría original 'categoria'

df_productos_nuevo = df_productos_nuevo.drop(columns=['categoria'])
df_productos_nuevo.head()
```

```python
# Recategorizando algunos productos (Bebidas Frias, Bebidas Calientes, Bebidas Alcohólicas)

# Definir las nuevas categorías y sus productos
nuevas_categorias = {
    'Bebidas_Frias': [
        'Pepsi 1.5L', 'Fanta Naranja 1.5L', 'Coca Cola 1.5L', 'Sprite 1.5L',
        'Agua Mineral 500ml', 'Energética Nitro 500ml', 'Jugo de Naranja 1L',
        'Jugo de Manzana 1L', 'Jugo en Polvo Limón', 'Jugo en Polvo Naranja'
    ],
    'Bebidas_Calientes': [
        'Yerba Mate Suave 1kg', 'Yerba Mate Intensa 1kg', 'Café Molido 250g',
        'Té Verde 20 saquitos', 'Té Negro 20 saquitos'
    ],
    'Bebidas_Alcoholicas': [
        'Cerveza Negra 1L', 'Cerveza Rubia 1L', 'Vino Blanco 750ml',
        'Fernet 750ml', 'Ron 700ml', 'Whisky 750ml', 'Vino Tinto Malbec 750ml',
        'Sidra 750ml', 'Vodka 700ml', 'Gin 700ml', 'Licor de Café 700ml'
    ]
}

# Función para asignar nueva categoría
def asignar_nueva_categoria(row):
    producto = row['nombre_producto']
    for categoria, productos in nuevas_categorias.items():
        if producto in productos:
            return categoria
    return row['categoria_corregida']  # Mantener la categoría original (Alimentos o Limpieza)

# Crear una copia del DataFrame original
df_productos_reclasificado = df_productos_nuevo.copy()

# Aplicar la nueva categorización
df_productos_reclasificado['nueva_categoria'] = df_productos_reclasificado.apply(asignar_nueva_categoria, axis=1)

# Mostrar un resumen de la nueva categorización
print("Resumen de la nueva categorización:")
print("\nNúmero de productos por categoría:")
print(df_productos_reclasificado['nueva_categoria'].value_counts())
```

```python
# Identificando outliers en el data frame detalle ventas

# Calculando Q1, Q3 e IQR para la variable importe
Q1 = df_detalle_ventas['importe'].quantile(0.25)
Q3 = df_detalle_ventas['importe'].quantile(0.75)
IQR = Q3 - Q1

# Calculando los límites para identificar outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Identificando los outliers
outliers = df_detalle_ventas[
    (df_detalle_ventas['importe'] < limite_inferior) | 
    (df_detalle_ventas['importe'] > limite_superior)
]

# Ordenando los outliers por importe de mayor a menor
outliers = outliers.sort_values('importe', ascending=False)

print("Número total de outliers encontrados:", len(outliers))
print("\nLímites para identificar outliers:")
print(f"Límite inferior: ${limite_inferior:.2f}")
print(f"Límite superior: ${limite_superior:.2f}")
print("\nDetalles de los outliers encontrados:")
print(outliers[['id_venta', 'id_producto', 'nombre_producto', 'cantidad', 'precio_unitario', 'importe']])
```

```python
# Calculando y graficando el importe total por nueva categoría

# Unir df_detalle_ventas con df_productos_reclasificado para obtener las categorías
df_ventas_categorias = pd.merge(
    df_detalle_ventas,
    df_productos_reclasificado[['id_producto', 'nueva_categoria']],
    on='id_producto',
    how='left'
)

# Calcular el importe total por categoría y ordenar de mayor a menor
importe_por_categoria = (df_ventas_categorias
                        .groupby('nueva_categoria')
                        ['importe']
                        .sum()
                        .sort_values(ascending=False)
                        .reset_index())

# Formatear los importes para las etiquetas
importe_por_categoria['importe_formato'] = importe_por_categoria['importe'].apply(lambda x: f'${x:,.0f}')

# Configurar el estilo y crear figura
plt.style.use('ggplot')
plt.figure(figsize=(12, 6))

# Crear el gráfico de barras
ax = sns.barplot(data=importe_por_categoria, 
                x='nueva_categoria', 
                y='importe', 
                palette='viridis')

# Personalizar título y etiquetas
plt.title("Importe Total de Ventas por Categoría de Producto", fontsize=16, pad=20, fontweight='bold')
plt.xlabel("Categoría de Producto", fontsize=12, fontweight='bold')
plt.ylabel("Importe Total ($)", fontsize=12, fontweight='bold')

# Rotar y ajustar las etiquetas del eje x
plt.xticks(rotation=45, ha='right')

# Añadir valores formateados sobre cada barra
for i, v in enumerate(importe_por_categoria['importe']):
    ax.text(i, v, importe_por_categoria['importe_formato'].iloc[i],
            ha='center', va='bottom', fontweight='bold')

# Formatear el eje y en miles
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x:,.0f}'))

# Ajustar los márgenes
plt.tight_layout()

# Añadir grid horizontal y remover líneas de borde
ax.yaxis.grid(True, linestyle='--', alpha=0.7)
sns.despine()

# Mostrar el gráfico
plt.show()

# Mostrar tabla con los totales y porcentajes
print("\nImporte total y porcentaje de ventas por categoría:")
total_ventas = importe_por_categoria['importe'].sum()
for _, row in importe_por_categoria.iterrows():
    porcentaje = (row['importe'] / total_ventas) * 100
    print(f"{row['nueva_categoria']}: {row['importe_formato']} ({porcentaje:.1f}%)")
```

```python
# Verificando que fecha_venta en df_ventas sea posterior a fecha_alta en df_clientes

# Primero, asegurémonos de que las fechas estén en el formato correcto
df_ventas['fecha_venta'] = pd.to_datetime(df_ventas['fecha_venta'])
df_clientes['fecha_alta'] = pd.to_datetime(df_clientes['fecha_alta'])

# Unir los DataFrames para comparar las fechas
df_ventas_merged = pd.merge(df_ventas, df_clientes[['id_cliente', 'fecha_alta']], 
                           on='id_cliente', how='left')

# Encontrar inconsistencias en fechas
inconsistencias_fecha = df_ventas_merged[df_ventas_merged['fecha_venta'] < df_ventas_merged['fecha_alta']]

# Mostrar resumen
print("=== Análisis de Consistencia de Fechas ===")
print(f"\nTotal de ventas analizadas: {len(df_ventas_merged)}")
print(f"Ventas con fecha anterior a fecha de alta del cliente: {len(inconsistencias_fecha)}")

if len(inconsistencias_fecha) > 0:
    print("\nDetalle de inconsistencias encontradas:")
    print(inconsistencias_fecha[['id_venta', 'id_cliente', 'fecha_venta', 'fecha_alta']].to_string())

    # Calcular la diferencia en días para las inconsistencias
    inconsistencias_fecha['diferencia_dias'] = (inconsistencias_fecha['fecha_alta'] - inconsistencias_fecha['fecha_venta']).dt.days

    print("\nEstadísticas de las diferencias de tiempo:")
    print(inconsistencias_fecha['diferencia_dias'].describe())
else:
    print("\n¡No se encontraron inconsistencias en las fechas!")
```

```python
# Calculando y graficando el importe en df_detalle_ventas por medio de pago en df_ventas

# Unir df_ventas con df_detalle_ventas para obtener el medio de pago asociado a cada venta
df_ventas_detalle = pd.merge(df_ventas, df_detalle_ventas, on='id_venta', how='left')

# Calcular el importe total por medio de pago
importe_por_medio_pago = (df_ventas_detalle
    .groupby('medio_pago')['importe'].sum()
).sort_values(ascending=False)

# Calcular el importe total para los porcentajes
total_importe = importe_por_medio_pago.sum()

# Configurar el estilo de ggplot para un look más profesional
plt.style.use('ggplot')

# Crear figura con tamaño personalizado
plt.figure(figsize=(8, 5))

# Definir una paleta de colores personalizada
colores = ['#2ecc71', '#3498db', '#9b59b6', '#f1c40f', '#e74c3c', '#1abc9c']

# Crear el gráfico de barras con colores personalizados
barras = plt.bar(importe_por_medio_pago.index, importe_por_medio_pago.values, 
                 color=colores[:len(importe_por_medio_pago)])

# Añadir valores y porcentajes sobre las barras
for i, v in enumerate(importe_por_medio_pago):
    porcentaje = (v / total_importe) * 100
    plt.text(i, v, f'${v:,.0f}\n({porcentaje:.1f}%)', 
             ha='center', va='bottom', fontweight='bold')

# Personalizar título y etiquetas
plt.title("Importe Total de Ventas por Medio de Pago", fontsize=16, pad=20, fontweight='bold')
plt.xlabel("Medio de Pago", fontsize=12, fontweight='bold')
plt.ylabel("Importe Total ($)", fontsize=12, fontweight='bold')

# Rotar etiquetas del eje x para mejor legibilidad
plt.xticks(rotation=45, ha='right')

# Añadir grid para mejor lectura
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

# Formatear eje y con formato de moneda
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x:,.0f}'))

# Ajustar márgenes
plt.tight_layout()

# Mostrar el gráfico
plt.show()
```

# Consolidando los datos en un solo data frame

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_clientes = pd.read_excel("clientes.xlsx")
df_productos_reclasificado = pd.read_excel("productos_reclasificados.xlsx")
df_ventas = pd.read_excel("ventas.xlsx")
df_detalle_ventas = pd.read_excel("detalle_ventas.xlsx")
```

```python
# 4. Fusiones para crear el DataFrame consolidado

# 4.1 Fusión de ventas y su detalle
print("4.1 Fusionando ventas con detalle_ventas...")
df_ventas_detalle = pd.merge(df_ventas, df_detalle_ventas, on='id_venta', how='left')
print(f"Dimensiones del DataFrame fusionado: {df_ventas_detalle.shape}")

# 4.2 Fusión con productos reclasificados
print("\n4.2 Fusionando con productos reclasificados...")
df_temp = pd.merge(df_ventas_detalle, df_productos_reclasificado, on='id_producto', how='left')
print(f"Dimensiones del DataFrame temporal: {df_temp.shape}")

# 4.3 Fusión con clientes para obtener el consolidado
print("\n4.3 Fusionando con clientes...")
df_consolidado = pd.merge(df_temp, df_clientes, on='id_cliente', how='left')
print(f"Dimensiones del DataFrame consolidado final: {df_consolidado.shape}")

# Mostrar las primeras filas del DataFrame consolidado
print("\nPrimeras filas del DataFrame consolidado:")

df_consolidado.head()
```

```python
# Limpiando df_consolidado eliminando columnas innecesarias

df_consolidado = df_consolidado.drop(columns=['nombre_cliente_y', 'email_y', 'nombre_producto_y', 'precio_unitario_y'])

```

```python
# Renombrando columnas en df_consolidado para mayor claridad
df_consolidado = df_consolidado.rename(columns={
    'nombre_cliente_x': 'nombre_cliente',
    'email_x': 'email',
    'nombre_producto_x': 'nombre_producto',
    'precio_unitario_x': 'precio_unitario'
})
df_consolidado.head()
```

```python
# Guardando df_consolidado en un archivo Excel
df_consolidado.to_excel("consolidado.xlsx", index=False)
```

```python
# Guardado df_consolidado en un archivo CSV
df_consolidado.to_csv("consolidado.csv", index=False)
```
